## ReACT Prompt

You are a Python coding assistant. Create a Python program that calculates the average of a list of student test scores.

Follow this ReACT-style process:

3. RUN: Execute the code using these scores: 85, 90, 78, 92, 88.

Requirements:

## Step 1: PLAN

The program will store the student test scores in a list. It will check whether the list is empty to prevent a division-by-zero error. If scores are available, it will calculate the average by adding the scores and dividing by the number of scores. The final average will be displayed with two decimal places.

```python
def calculate_average(scores):
    if len(scores) == 0:
        return "No scores available."

    average = sum(scores) / len(scores)
    return round(average, 2)


scores = [85, 90, 78, 92, 88]

result = calculate_average(scores)

print("Student Scores:", scores)
print("Average Score:", result)
```

## Step 3: RUN and OBSERVE

## Step 4: FIX

I updated the code so the average always displays exactly two decimal places. I also tested an empty list to make sure the program handles it without causing an error.

```python
def calculate_average(scores):
    if len(scores) == 0:
        return None

    return sum(scores) / len(scores)


# Test 1: Normal list
scores = [85, 90, 78, 92, 88]
result = calculate_average(scores)

if result is not None:
    print("Student Scores:", scores)
    print(f"Average Score: {result:.2f}")
else:
    print("No scores available.")


# Test 2: Empty list
empty_scores = []
empty_result = calculate_average(empty_scores)

print("\nEmpty List Test:")

if empty_result is not None:
    print(f"Average Score: {empty_result:.2f}")
else:
    print("No scores available.")
```

## Step 5: Final Observation

After fixing the formatting, the program successfully displayed the average as 86.60. I also tested the program with an empty list, and it returned "No scores available" instead of causing an error.
